```python
from ultralytics import YOLO
model = YOLO("yolo11n.pt")
```

```python
import zipfile
```

```python
zipfile.ZipFile("/content/Helmet_Detection.v7i.yolov11.zip").extractall()
```

```python
model.train(data='/content/Helmet Detection.v7i.yolov11/data.yaml',epochs=20)
```

```python
metrics = model.val(
    data="/content/Helmet Detection.v7i.yolov11/data.yaml",
    split="test",
    imgsz=640,
    batch=16,
    device=0,
    plots=True
)

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```
