```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
import zipfile
zipfile.ZipFile("/content/wrong-way.v1i.yolov11.zip").extractall()
```

```python
model.train(data='/content/wrong-way.v1i.yolov11/data.yaml',epochs=20)
```

```python
metrics = model.val(
    data="/content/wrong-way.v1i.yolov11/data.yaml",
    split="test",
    imgsz=640,
    batch=16,
    device=0,
    plots=True,
    conf=0.25 # Iss se model ziada detections pick karega
)
```
